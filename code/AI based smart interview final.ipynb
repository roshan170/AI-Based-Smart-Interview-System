```python
import PyPDF2
def extract_text(pdf_path):
    text = ""
    with open(pdf_path,'rb') as file:
        reader = PyPDF2.PdfReader(file)
        for page in reader.pages:
            text += page.extract_text()
    return text
```

```python
resume_text = extract_text("Roshan_Data_Science.pdf")
print(resume_text[:500])
```

```python
skills_list=["python","machine learning","sql","data analysis","deep learning","cloud","docker","kubernets","excel","power bi","tableau","nlp"]
def extract_skills(text):
    text=text.lower()
    found=[]
    for skill in skills_list:
        if skill in text:
            found.append(skill)
    return found
```

```python
skills = extract_skills(resume_text)
print("Detected Skills:", skills)
```

```python
def detect_job_role(skills):
    if "machine learning" in skills or "nlp" in skills:
        return "Machine Learning Engineer"
    elif "sql" in skills and "excel" in skills:
        return "Data Analyst"
    elif "python" in skills and "deep learning" in skills:
        return "AI Engineer"
    else:
        return "Software Developer"
```

```python
job_role = detect_job_role(skills)
print("Predicted Job Role:", job_role)
```

```python
data = pd.read_csv('fer2013.csv')
pixels = data['pixels'].tolist()
faces = []
for pixel_sequence in pixels:
    face = np.array(pixel_sequence.split(' '), dtype='float32')
    face = face.reshape(48,48)
    faces.append(face)
faces = np.array(faces)
faces = np.expand_dims(faces,-1)
faces = faces/255.0
```

```python
import tensorflow as tf
import keras
import h5py
```

```python
from tensorflow.keras.utils import to_categorical
emotions = pd.get_dummies(data['emotion']).values
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D,MaxPooling2D,Flatten,Dense,Dropout
```

```python
model = Sequential()
model.add(Conv2D(32,(3,3),activation='relu',input_shape=(48,48,1)))
model.add(MaxPooling2D(2,2))
model.add(Conv2D(64,(3,3),activation='relu'))
model.add(MaxPooling2D(2,2))
model.add(Conv2D(128,(3,3),activation='relu'))
model.add(Flatten())
model.add(Dense(128,activation='relu'))
model.add(Dropout(0.5))
model.add(Dense(7,activation='softmax'))
```

```python
model.compile(
optimizer='adam',
loss='categorical_crossentropy',
metrics=['accuracy']
)
```

```python
model.fit(
faces,
emotions,
epochs=10,
batch_size=64,
validation_split=0.2
)
```

```python
model.save("emotion_model.hdf5")
```

```python
import threading
import sounddevice as sd
from scipy.io.wavfile import write
```

```python
fs=44100
record_seconds=15
audio_data=[]
```

```python
def record_audio():
    global audio_data
    print("Speak your answer...")
    audio_data = sd.rec(int(record_seconds * fs), samplerate=fs, channels=1,dtype='int16')
    sd.wait()
    write("answer.wav", fs, audio_data)
    print("Audio Recorded")
```

```python
detected_emotion=[]
def emotion_detection():
    cap = cv2.VideoCapture(0)
    print("Camera on(press 'q'to stop)")
    
    while True:
        ret,frame=cap.read()
        if not ret:
            break
        gray=cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
        faces=face_cascade.detectMultiScale(gray,1.3,5)
        for (x,y,w,h) in faces:
            face=gray[y:y+h,x:x+w]
            face=cv2.resize(face,(48,48))
            face=face/255
            face=np.reshape(face,(1,48,48,1))
            prediction=emotion_model.predict(face)
            emotion=emotion_labels[np.argmax(prediction)]
            cv2.putText(frame,"Emotion:"+emotion,(x,y-10),
                    cv2.FONT_HERSHEY_SIMPLEX,0.8,(0,255,0),2)
        cv2.imshow("Interview Camera",frame)
        if cv2.waitKey(1)&0xFF==ord('q'):
            break
    cap.release()
    cv2.destroyAllWindows()
```

```python
audio_thread=threading.Thread(target=record_audio)
video_thread=threading.Thread(target=emotion_detection)
audio_thread.start()
video_thread.start()
audio_thread.join()
video_thread.join()
```
